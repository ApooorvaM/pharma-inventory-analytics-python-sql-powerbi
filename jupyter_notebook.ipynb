Feature Engineering: Adding new columns to understand the avg demand and how many months of inventory coverage we have according to avg demand.

```python
sales['avg_monthly_demand'] = sales.groupby(by = ['product_id','year'])['sales_quantity'].transform('mean')
sales.head()
```

```python
sales['std_dev_monthly_demand'] = sales.groupby(by = ['product_id','year'])['sales_quantity'].transform('std')
sales['coeff_of_variability'] = sales['std_dev_monthly_demand']/sales['avg_monthly_demand']*100.0
sales.head()
```

```python
sales['inventory_coverage'] = sales['inventory_quantity']/sales['avg_monthly_demand']
```

```python
sales = pd.merge(sales, product[['product_id','storage_cost_per_unit_month','price_per_unit']], how = 'left', on = 'product_id')
sales.head()
```

```python
# Calculating cost for storing/holding the inventory and the value of inventory

sales['inventory_value'] = sales['inventory_quantity']*sales['price_per_unit']
sales['holding_cost'] = sales['inventory_quantity']*sales['storage_cost_per_unit_month']
sales.head()
```

Calculating required inventory for each product for each month using lead time and safety stock

```python
sales = pd.merge(sales,product[['product_id','lead_time_months','safety_stock_months','shelf_life_months']], how ='left', on = 'product_id' )
```

```python
sales.head()
```

```python
sales ['inventory_req'] = sales['avg_monthly_demand'] * (sales['lead_time_months'] + sales['safety_stock_months'])
```

```python
import urllib.parse
from sqlalchemy import create_engine

# Database settings
host = r"localhost\SQLEXPRESS"
database = "healthcare_inventory"

# Build connection string for Windows Authentication
connection_string = (
    f"DRIVER={{ODBC Driver 17 for SQL Server}};"
    f"SERVER={host};"
    f"DATABASE={database};"
    f"Trusted_Connection=yes;"
)

params = urllib.parse.quote_plus(connection_string)
engine = create_engine(f"mssql+pyodbc:///?odbc_connect={params}")

# ---------------------------------------------------------
# Upload your 3 DataFrames to SQL Server
# ---------------------------------------------------------

# 1. Upload Sales Table
sales.to_sql("sales", engine, if_exists="append", index=False)

# 2. Upload Product Table
product.to_sql("product", engine, if_exists="append", index=False)

# 3. Upload Finance Table
finance.to_sql("finance", engine, if_exists="append", index=False)

print("All 3 tables (sales, product, finance) uploaded successfully to 'healthcare_inventory'!")
```

```python
def stock_status (sales):
    if sales['inventory_quantity'] < sales['inventory_req']:
       return 'understock'
    elif sales['inventory_req'] <= sales['inventory_quantity'] < 1.5 * sales['inventory_req']:
       return 'healthy'
    else:
       return 'overstock'
        
```

```python
sales['stock_status'] = sales.apply(stock_status, axis = 1)
sales[['product_id','inventory_quantity','inventory_req','stock_status']].head()
```

```python
# WARNING: This deletes the old SQL table and creates a brand new one
sales.to_sql("sales", engine, if_exists="replace", index=False)
```
